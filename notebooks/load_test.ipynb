```python
# load_data,py serves to collate and clean the data from the 05/06 to 25/26 season
# ENSURE CMD IS ONLY RUN FROM THE ROOT DIR
import pandas as pd
from pathlib import Path

dir = Path("data/cs/")
histories = []
for file in dir.iterdir():
    df = pd.read_csv(file, low_memory=False)
    series = pd.Series(Path(file).stem, index=range(380))
    series = series.astype(dtype="int64")
    series = series.apply(lambda x: x // 100 + 2000)
    series.name = "season"
    df = pd.concat([df,series], axis=1)
    histories.append(df)
df = pd.concat(histories)
df["Date"] = pd.to_datetime(df["Date"], format="mixed", dayfirst=True)
# df = df.reset_index(drop=True) # possible to remove this line-> allow to find different seasons-> reset form...
series = pd.Series(False, index=range(380))
series.name = "HP"
awaySeries = series.copy()
awaySeries.name = "AP"
df = pd.concat([df,series,awaySeries], axis=1)

df.to_csv("data/csHistory.csv", index=False)
```

```python
teams = set()
curSeason = 2005
for row in df.itertuples():
    if row["season"] == 2005:
        teams.add(row["HomeTeam"])
        teams.add(row["AwayTeam"])
        continue
    else:
        if row["HomeTeam"] in teams or row["AwayTeam"] in teams:
            continue 


```
